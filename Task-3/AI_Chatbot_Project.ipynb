```python
from transformers import pipeline

chatbot = pipeline("text-generation", model="gpt2")
```

```python
response = chatbot("What is Artificial Intelligence?", max_length=50)
print(response)
```

```python
print(response[0]['generated_text'])
```

```python
while True:
    user_input = input("You: ")
    
    if user_input.lower() == "exit":
        print("Chatbot: Bye 👋")
        break
    
    response = chatbot(user_input, max_length=100)
    print("Chatbot:", response[0]['generated_text'])
```

```python
from langchain_community.llms import HuggingFacePipeline
```

```python
from transformers import pipeline

pipe = pipeline("text-generation", model="gpt2")

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
llm.invoke("What is Artificial Intelligence?")
```

```python
chat_history = []

while True:
    user_input = input("You: ")
    
    if user_input.lower() == "exit":
        print("Chatbot: Bye 👋")
        break
    
    chat_history.append("User: " + user_input)
    
    prompt = "You are a helpful and polite AI assistant.\n" + "\n".join(chat_history[-4:]) + "\nChatbot:"
    
    response = llm.invoke(prompt)
    
    reply = response.replace(prompt, "").strip().split("\n")[0]
    
    print("Chatbot:", reply)
    
    chat_history.append("Chatbot: " + reply)
```
